# Etude de quantization 1

```python
def quantize(float_value: np.ndarray, min_range, max_range, zero = 0):
    b = 8
    S = (max_range - min_range) / (2**b - 1)
    quantized = (float_value / S).astype(int) + zero
    return quantized
```

```python
def to_float(uint_values: np.ndarray, min_range, max_range):
    b = 8
    S = (max_range - min_range) / (2**b - 1)
    unquantized = uint_values.astype(np.float32) * S
    return unquantized
```

```python
def quantization_error_stats(values, min, max):
    quantized = quantize(values, min, max)
    unquantized = to_float(quantized, min, max)
    error = values - unquantized
    return error
```

```python
def display_error_histogram(values, min, max): #TODO, check values
    import matplotlib.pyplot as plt

    error = quantization_error_stats(values, min, max)
    plt.hist(error, bins=50)
    plt.title("Quantization Error Histogram")
    plt.xlabel("Error")
    plt.ylabel("Fréquence")
    plt.show()
```

## Situation 1

```python
array = np.random.uniform(0, 0.5, 1000).astype(np.float32)
min_range = 0
max_range = 0.5

error = quantization_error_stats(array, min_range, max_range)
display_error_histogram(error, min_range, max_range)
```
